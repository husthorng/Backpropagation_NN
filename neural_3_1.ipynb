```python
import csv
import numpy as np
import pandas as pd
url = "https://raw.githubusercontent.com/husthorng/hello-world/neural/trainingdata.csv"
import numpy as np
def nonlin(x,deriv=False):
    A=1/(1+np.exp(-x))
    if(deriv==True):
        return (A*(1-A))
    return A
# print(nonlin(0,1))

numTag=1   #
hidden_node=3

records = pd.read_csv(url,encoding='utf-8')
#print(records.shape)
inputD=records.iloc[0:records.shape[0],0:records.shape[1]-numTag].values
#print(inputD.shape)
tagetD=records.iloc[:,list(records.shape[1]-np.arange(numTag,0,-1))].values
print(inputD)
print(tagetD)
np.random.seed(1)
syn0=2*np.random.random((inputD.shape[1],hidden_node))-1
syn1=2*np.random.random((hidden_node,numTag))-1
# print(syn0)
# print(syn1)
for j in range(1200):
    X=inputD
    L0=nonlin(np.dot(inputD,syn0))
    L1=nonlin(np.dot(L0,syn1))
    #print(L1)
    L1_error=tagetD-L1
    if(j % 10)==0:
        print("Error"+str(np.mean(np.abs(L1_error))))
    #print(L1_error)
    L1_delta=L1_error*nonlin(L1,deriv=True)
    #print(L1_delta)
    L0_error=L1_delta.dot(syn1.T)
    L0_delta=L0_error*nonlin(L0,deriv=True)
    syn1 += L0.T.dot(L1_delta)
    syn0 += inputD.T.dot(L0_delta)
print("output after Training:")
# np.savetxt("syn0.csv", syn0, delimiter=",")
# np.savetxt("syn1.csv", syn1, delimiter=",")
#print(syn0)
df = pd.DataFrame(syn0)
df.to_csv("syn0.csv")
df = pd.DataFrame(syn1)
df.to_csv("syn1.csv")
print(L1)
print(syn0)
print(syn1)
```

```python
import numpy as np
import pandas as pd

# ==========================================
# Google Sheet ID
# ==========================================

SHEET_ID = "17SfXcwlQCx2O_2XStirVF9j8Q4M5b9PWy4Kb7WOMW9Q"


# ==========================================
# 開啟 Google Sheet
# ==========================================

sh = gc.open_by_key(SHEET_ID)

worksheet = sh.worksheet("DATA")


# ==========================================
# 讀取全部資料
# ==========================================

data = worksheet.get_all_records()

df = pd.DataFrame(data)

print("Google Sheet 資料：")

display(df)
```

```python
# ==========================================
# Neural Network Input
# ==========================================

X = df[["X1", "X2", "X3"]].values.astype(float)


# ==========================================
# Neural Network Output
# ==========================================

y = df[["Y"]].values.astype(float)


print("X =")

print(X)


print("\ny =")

print(y)


print("\nX shape =", X.shape)

print("y shape =", y.shape)
```

```python
# ==========================================
# Neural Network
# 3 Input
# 3 Hidden
# 1 Output
# ==========================================

import numpy as np


# ==========================================
# 固定亂數
# ==========================================

np.random.seed(1)


# ==========================================
# Sigmoid
# ==========================================

def sigmoid(x):
    x = np.clip(x, -500, 500)
    return 1 / (1 + np.exp(-x))


# ==========================================
# Sigmoid 微分
# ==========================================

def sigmoid_derivative(x):

    return x * (1 - x)


# ==========================================
# 初始化權重
# ==========================================

# Input(3) → Hidden(3)

W0 = np.random.uniform(
    -1,
    1,
    (3, 3)
)


# Hidden(3) → Output(1)

W1 = np.random.uniform(
    -1,
    1,
    (3, 1)
)


# ==========================================
# Bias
# ==========================================

# Hidden Layer Bias

b0 = np.zeros((1, 3))


# Output Layer Bias

b1 = np.zeros((1, 1))


# ==========================================
# Training Parameter
# ==========================================

learning_rate = 0.5

epochs = 50000


# ==========================================
# Training
# ==========================================

for epoch in range(epochs):


    # --------------------------------------
    # Forward Propagation
    # --------------------------------------


    # Input → Hidden

    hidden_input = np.dot(X, W0) + b0

    hidden_output = sigmoid(
        hidden_input
    )


    # Hidden → Output

    output_input = np.dot(
        hidden_output,
        W1
    ) + b1


    output = sigmoid(
        output_input
    )


    # --------------------------------------
    # Error
    # --------------------------------------

    error = y - output


    # --------------------------------------
    # Back Propagation
    # --------------------------------------


    # Output Delta

    output_delta = (
        error *
        sigmoid_derivative(output)
    )


    # Hidden Error

    hidden_error = np.dot(
        output_delta,
        W1.T
    )


    # Hidden Delta

    hidden_delta = (
        hidden_error *
        sigmoid_derivative(
            hidden_output
        )
    )


    # --------------------------------------
    # Update W1
    # --------------------------------------

    W1 += (
        learning_rate *
        np.dot(
            hidden_output.T,
            output_delta
        )
    )


    # --------------------------------------
    # Update W0
    # --------------------------------------

    W0 += (
        learning_rate *
        np.dot(
            X.T,
            hidden_delta
        )
    )


    # --------------------------------------
    # Update Bias
    # --------------------------------------

    b1 += (
        learning_rate *
        np.sum(
            output_delta,
            axis=0,
            keepdims=True
        )
    )


    b0 += (
        learning_rate *
        np.sum(
            hidden_delta,
            axis=0,
            keepdims=True
        )
    )


    # --------------------------------------
    # 顯示 Loss
    # --------------------------------------

    if epoch % 5000 == 0:

        loss = np.mean(
            error ** 2
        )

        print(
            "Epoch:",
            epoch,
            "Loss:",
            loss
        )


print("\nTraining Complete")
```

```python
# ==========================================
# Prediction
# ==========================================


hidden = sigmoid(
    np.dot(X, W0) + b0
)


output = sigmoid(
    np.dot(hidden, W1) + b1
)


# ==========================================
# Binary Result
# ==========================================

binary_output = (
    output >= 0.5
).astype(int)


# ==========================================
# 建立結果表
# ==========================================

result = pd.DataFrame()


result["X1"] = X[:, 0]

result["X2"] = X[:, 1]

result["X3"] = X[:, 2]


result["Expected"] = y.flatten()


result["Prediction"] = output.flatten()


result["Binary"] = binary_output.flatten()


display(result)
```

```python
//neural_networks_3_1_arduino.ino
// https://drive.google.com/file/d/1L52F7ktVJUKTEYK5yJQf8X2ryXfS80F1/view?usp=drive_link
// ==========================================
// ESP32 Neural Network
// Architecture:
// Input  = 3
// Hidden = 3
// Output = 1
// ==========================================

#include <Arduino.h>
#include <math.h>


// ==========================================
// Neural Network Weight
// 請貼入 Colab 訓練後的數值
// ==========================================


// Input → Hidden

const float W0[3][3] = {
  {1.59200208f, 2.85073462f, -2.00639241f},
  {4.24092945f, -7.17099873f, -4.93668278f},
  {-7.07100716f, 4.80771046f, -4.66530027f}
};

const float B0[3] = {
-1.43628832f, -1.92109311f, 0.98531392f};

const float W1[3] = {
10.60429151f, 10.60915928f, -5.75757664f};

const float B1 = -4.89308433f;


// ==========================================
// Sigmoid
// ==========================================

float sigmoid(float x)
{
  return 1.0f /
         (1.0f + exp(-x));
}


// ==========================================
// Neural Network Prediction
// ==========================================

float neuralNetwork(
  float X1,
  float X2,
  float X3
)
{

  // --------------------------------------
  // Input Array
  // --------------------------------------

  float input[3] = {

    X1,

    X2,

    X3

  };


  // --------------------------------------
  // Hidden Layer
  // --------------------------------------

  float hidden[3];


  for (int j = 0; j < 3; j++)
  {

    float sum = B0[j];


    for (int i = 0; i < 3; i++)
    {

      sum += input[i] * W0[i][j];

    }


    hidden[j] = sigmoid(sum);

  }


  // --------------------------------------
  // Output Layer
  // --------------------------------------

  float outputSum = B1;


  for (int i = 0; i < 3; i++)
  {

    outputSum += hidden[i] * W1[i];

  }


  float output = sigmoid(outputSum);


  return output;

}


// ==========================================
// Setup
// ==========================================

void setup()
{

  Serial.begin(115200);


  Serial.println();

  Serial.println(
    "ESP32 Neural Network Start"
  );


  // --------------------------------------
  // Test Input
  // --------------------------------------

  float X1 = 1;

  float X2 = 0;

  float X3 = 1;


  // --------------------------------------
  // Neural Network Prediction
  // --------------------------------------

  float result = neuralNetwork(
    X1,
    X2,
    X3
  );


  Serial.print(
    "Neural Network Output = "
  );


  Serial.println(
    result,
    6
  );


  // --------------------------------------
  // Binary Output
  // --------------------------------------

  if (result >= 0.5)
  {

    Serial.println(
      "Result = 1"
    );

  }

  else
  {

    Serial.println(
      "Result = 0"
    );

  }

}


// ==========================================
// Loop
// ==========================================

void loop()
{

}
```
